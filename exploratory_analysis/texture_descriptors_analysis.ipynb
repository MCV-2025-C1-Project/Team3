```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import re
```

```python
# Load your CSV file
df = pd.read_csv("/home/gerardasbert/Desktop/Master/C1/projecte/Team3/results/dev_scores_bon_noise_removal_LBP.csv")

# Extract the LBP method name from the descriptor string using regex
df['method'] = df['descriptor'].apply(
    lambda x: re.search(r"'method': '([^']+)'", x).group(1) if re.search(r"'method': '([^']+)'", x) else None
)

# Plot violin plot for mean_mAP by method
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='method', y='mean_mAP', inner='box',)

plt.title("Distribution of mean_mAP by LBP Method")
plt.xlabel("LBP Method")
plt.ylabel("mean_mAP")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Load your CSV file
df = pd.read_csv("/home/gerardasbert/Desktop/Master/C1/projecte/Team3/results/dev_scores_bon_noise_removal_LBP.csv")

# Extract the R value (radius) from the descriptor string using regex
df['R'] = df['descriptor'].apply(
    lambda x: int(re.search(r"'R': (\d+)", x).group(1)) if re.search(r"'R': (\d+)", x) else None
)

# Plot violin plot for mean_mAP by R
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='R', y='mean_mAP', inner='box', palette='Set2')

plt.title("Distribution of mean_mAP by LBP Radius (R)")
plt.xlabel("LBP Radius (R)")
plt.ylabel("mean_mAP")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Load your CSV file
df = pd.read_csv("/home/gerardasbert/Desktop/Master/C1/projecte/Team3/results/dev_scores_bon_noiseremoval_DCT_set1.csv")

# Extract the R value (radius) from the descriptor string using regex
df['block_size'] = df['descriptor'].apply(
    lambda x: int(re.search(r"'block_size': (\d+)", x).group(1)) if re.search(r"'block_size': (\d+)", x) else None
)

# Plot violin plot for mean_mAP by block_size
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='block_size', y='mean_mAP', inner='box', bw_adjust=0.1, cut=0, palette='coolwarm')

plt.title("Distribution of mean_mAP by DCT block size")
plt.xlabel("DCT Block Size")
plt.ylabel("mean_mAP")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
